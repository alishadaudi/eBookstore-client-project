```python
# Ensure pandas is imported and url is defined (assuming previous cells M5ymeYqyIr3N and VguFioTsJiRk have been run)
import pandas
url = 'https://docs.google.com/spreadsheets/d/1XSBE7VwTJ-xB300r2JAwZesaAbGTLvMgS4qRxXfDjys/pub?gid=0&single=true&output=csv'

# Code snippet 3: Importing the dataset from the URL using pandas package
data = pandas.read_csv(url)

# Code snippet 9: Dropping  'Title', 'Middle Name', 'Suffix', 'Street Address2' because of missing more than 20% data
data80 = data.drop(['Title', 'Middle Name', 'Suffix', 'Street Address2'], axis=1)

# Code snippet 11: Dropping noise variables - 'Customer ID', 'First Name', 'Last Name','Street Address1'
predictors = data80.drop(['Customer ID', 'First Name', 'Last Name','Street Address1'], axis=1)

# Display the predictors to confirm it's defined
print(predictors.columns)
```

```python
# Code snippet 12
# Displaying predictors size/shape
predictors.shape
```

```python
import pandas as pd
from datetime import date

# Code snippet 13
# Convert 'Birth Date' to datetime objects
predictors['Birth Date'] = pd.to_datetime(predictors['Birth Date'])

# Calculate Age from 'Birth Date'
def calculate_age(born):
    today = date.today()
    return today.year - born.year - ((today.month, today.day) < (born.month, born.day))

predictors['Age'] = predictors['Birth Date'].apply(calculate_age)

# Remove 'Birth Date' column as it's redundant
predictors = predictors.drop('Birth Date', axis=1)

# Display the updated predictors DataFrame with the new 'Age' column
predictors.head()
```

```python
# Code snippet 14
# Split 'City-ZipCode-State' into 'City', 'Zip Code', and 'State'
predictors[['City', 'Zip Code', 'State']] = predictors['City-ZipCode-State'].str.split('-', expand=True)

# Display the predictors DataFrame with the new columns
predictors.head()
```

```python
# Code snippet 15
# Drop the original 'City-ZipCode-State' column
predictors = predictors.drop('City-ZipCode-State', axis=1)

# Display the updated columns of the predictors DataFrame
predictors.columns
```
