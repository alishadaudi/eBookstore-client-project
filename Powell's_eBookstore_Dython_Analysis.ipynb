```python
# Code snippet 9
# Dropping  'Title', 'Middle Name', 'Suffix', 'Street Address2' because of missing more than 20% data
# Quantitative Data Preparation
data80 = data.drop(['Title', 'Middle Name', 'Suffix', 'Street Address2'], axis=1)

# Display the data80 columns
data80.columns
```

```python
# Code snippet 11
# Dropping noise variables - 'Customer ID', 'First Name', 'Last Name','Street Address1'
# Qualitative Data Preparation
predictors = data80.drop(['Customer ID', 'First Name', 'Last Name','Street Address1'], axis=1)

# Display the predictors
predictors.columns
```

```python
# Code snippet 12
# Displaying predictors size/shape
predictors.shape
```

```python
# Code snippet 13
from datetime import date
import pandas

def calculate_age(born):
    today = date.today()
    return today.year - born.year - ((today.month, today.day) < (born.month, born.day))

# Convert 'Birth Date' to datetime objects, handling potential future dates
def correct_birth_year(dt):
    if dt.year > date.today().year:
        # Assume 20xx dates are actually 19xx if they are in the future
        return dt.replace(year=dt.year - 100)
    return dt

predictors['Birth Date'] = pandas.to_datetime(predictors['Birth Date'], errors='coerce')
predictors['Birth Date'] = predictors['Birth Date'].apply(correct_birth_year)

# Calculate Age from 'Birth Date'
predictors['Age'] = predictors['Birth Date'].apply(calculate_age)

# Display the updated predictors DataFrame with the new 'Age' column
predictors.head()
```

```python
# Code snippet 14
# The 'City-ZipCode-State' column has already been split into 'City', 'Zip Code', and 'State',
# and the original column has been dropped in a previous step or execution.
# Display the predictors DataFrame with the new columns and Age column
predictors.head()
```

```python
# Code snippet 15
# The original 'City-ZipCode-State' column has already been dropped in a previous step.

# Display the updated columns of the predictors DataFrame
predictors.columns
```

```python
# Code snippet 16
# Remove 'Birth Date' as 'Age' has already been calculated
# Use errors='ignore' to prevent KeyError if the column is already dropped
predictors = predictors.drop('Birth Date', axis=1, errors='ignore')

# Display the updated predictors DataFrame to confirm the change
predictors.head()
```

```python
# Code snippet 17
# Display the updated predictors DataFrame
predictors.head()
```

```python
# Code snippet 18
# Importing relevant statistical tests/modules from Dython
from dython.nominal import associations
```

```python
# Code snippet 19
import pandas
# Running the associations function on predictor variables
# The 'Birth Date' column has already been removed as 'Age' has been derived from it.
# No need to drop it again.
predictors_for_assoc = predictors.copy() # Create a copy to avoid modifying the original predictors DataFrame if further changes are made
assoc = associations(predictors_for_assoc)
```

```python
# Code snippet 20
# Storing the association values in a matrix
assoc_matrix = assoc['corr']
```

```python
# Code snippet 21
# Setting the target variable
target = 'Avg Monthly Spend'
```

```python
# Code snippet 22
# Listing the most impacting/associated variables with the target variable
assoc_matrix[target].abs().sort_values(ascending=False)
```
